```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
#Ruta desde la raiz del proyecto
ruta_excel = Path("../data/raw/base_datos.xlsx")

#Leer el archivo
df = pd.read_excel(ruta_excel)

#Mostrar las primeras filas
display(df.head())
```

```python
# Matriz de correlación de las columnas numéricas
df_numerico = df.select_dtypes(include=np.number)
matriz_corr = df_numerico.corr()

display(matriz_corr.round(2))

plt.figure(figsize=(14, 10))
sns.heatmap(matriz_corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Matriz de correlación")
plt.tight_layout()
plt.show()

# Variables con correlación absoluta menor que 0.7 respecto a Pago_atiempo
objetivo = "Pago_atiempo"

if objetivo in matriz_corr.columns:
    correlaciones = (
        matriz_corr[objetivo]
        .drop(objetivo)
        .dropna()
        .to_frame(name="correlacion")
    )
    correlaciones["correlacion_abs"] = correlaciones["correlacion"].abs()

    variables_menor_07 = (
        correlaciones[correlaciones["correlacion_abs"] < 0.7]
        .sort_values("correlacion_abs")
    )

    display(variables_menor_07)
else:
    print(f"'{objetivo}' no es numérica o no está disponible para calcular correlaciones.")
```

```python
columnas_numericas = df.select_dtypes(include=np.number).columns
n_columnas = 4
n_filas = (len(columnas_numericas) + n_columnas - 1) // n_columnas

fig, axes = plt.subplots(n_filas, n_columnas, figsize=(16, 4 * n_filas))
axes = np.array(axes).reshape(-1)

for ax, columna in zip(axes, columnas_numericas):
    sns.boxplot(y=df[columna], ax=ax)
    ax.set_title(columna)
    ax.set_xlabel("")

for ax in axes[len(columnas_numericas):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```
